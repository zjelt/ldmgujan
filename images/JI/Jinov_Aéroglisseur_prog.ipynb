# <div style= "text-align:center ; color:red">Aéroglisseur</div>

??? tip En savoir plus sur l'aéroglisseur
![sc](https://lyceedelamergujan.fr/SI/AERO/hover.png)

Un aéroglisseur utilise un coussin d'air en mouvement pour créer un petit espace entre le véhicule et la surface.

Ceci supprime la friction. L'aéroglisseur peut avancer en soufflant de l'air grâce à un ou plusieurs moteurs arrière.

![sc](https://lyceedelamergujan.fr/SI/AERO/hover0.png)

Les tables de hockey sur air utilisent le même principe pour éliminer les frottements. 

L'air est gonflé par un ventilateur afin de créer un petit espace d'air entre les disques et la table, supprimant ainsi les frottements. 

Les disques peuvent flotter et rebondir à grande vitesse et mettent longtemps avant de s'arrêter.

## <div class = "alert alert-block alert-danger" style= "text-align : center">`Carte réceptrice` </div>
[Retour au sommaire](#s)

!!! note Principe de fonctionnement de l'aéroglisseur
![sc](https://lyceedelamergujan.fr/SI/AERO/prin.png)
!!!

!!! note Une carte `réceptrice` programmable connecté à une carte de contrôle pilotera :
- le **moteur à coussin d'air**
- le **moteur propulseur**
- et le **servomoteur** relié au gouvernail
![sc](https://lyceedelamergujan.fr/SI/AERO/branchement.png)

La carte `réceptrice` :
- **reçoit par radio** les signaux de commande 
- et **pilote** :
    - le servomoteur sur la broche P1,
    - le ventilateur à coussin d'air sur la broche P0 et
    - le ventilateur propulseur sur la broche P2.

Les signaux sont amplifiés par la carte de contrôle et envoyés aux moteurs et au servo.

??? tip En savoir plus sur les hélices
<u>**Principe général**</u>

Une pale d’hélice fonctionne comme une aile :
- En tournant, elle dévie l’air vers l’arrière.
- Cette déviation crée une poussée dans la direction opposée (principe d’action–réaction).

<u>**Sens de rotation**</u>

Il existe généralement deux types d’hélices :
- **Hélice à pas droit** (rotation horaire)
    - Elle pousse l’air efficacement lorsqu’elle tourne dans le sens des aiguilles d’une montre (vue de face).
- **Hélice à pas gauche** (rotation antihoraire)
    - Elle est conçue pour tourner dans le sens inverse.

Si une hélice est montée sur un moteur qui tourne dans le mauvais sens, elle :
- brasse l’air au lieu de le pousser,
- produit peu ou pas de poussée.

**Comment reconnaître le bon sens ?**
- En regardant l’hélice de face, les pales doivent être inclinées vers l’arrière par rapport au sens de rotation.
- Le bord d’attaque (le plus épais) doit rencontrer l’air en premier.
???

!!! danger Avertissement : 

**Attention ! Ne touchez pas aux hélices lorsque l’appareil est sous tension.** 

En cas de problème, débranchez immédiatement la batterie. 
![sc](https://lyceedelamergujan.fr/SI/AERO/test4.png)

Variable | Type | Minimum | Milieu | Maximum
 :- | :- | :- | :- | :- 
`On` | Binaire | 0 |  | 1

            - le `tangage` : rotation autour de l'axe `x`(`pitch`)

                - lorsque l'axe `x` et l'axe `y` sont parallèles au sol alors `pitch = 0` et `roll = 0`
